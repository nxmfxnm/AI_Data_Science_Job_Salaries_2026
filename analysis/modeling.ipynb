```python
# ==========================================
# Random Forest Regression
# ==========================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer

from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score


# ==========================================
# 1. กำหนดตัวแปร X และ Y
# ==========================================

X = df[
    [
        "years_experience",
        "education_level",
        "experience_level",
        "company_size"
    ]
]

y = df["salary_usd"]


# ==========================================
# 2. แบ่งข้อมูล Train / Test
# ==========================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data :", X_test.shape)


# ==========================================
# 3. กำหนดประเภทตัวแปร
# ==========================================

numeric_features = [
    "years_experience"
]

categorical_features = [
    "education_level",
    "experience_level",
    "company_size"
]


# ==========================================
# 4. จัดการ Missing Values
# ==========================================

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ]
)

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(handle_unknown="ignore")
        )
    ]
)


# ==========================================
# 5. รวมการเตรียมข้อมูล
# ==========================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_transformer,
            numeric_features
        ),
        (
            "categorical",
            categorical_transformer,
            categorical_features
        )
    ]
)


# ==========================================
# 6. สร้าง Random Forest
# ==========================================

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)


# ==========================================
# 7. สร้าง Pipeline
# ==========================================

model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("random_forest", rf_model)
    ]
)


# ==========================================
# 8. Train Model
# ==========================================

print("\nกำลังสร้าง Random Forest...")

model.fit(X_train, y_train)

print("สร้าง Random Forest สำเร็จ")


# ==========================================
# 9. ทำนายข้อมูล Test
# ==========================================

y_pred = model.predict(X_test)


# ==========================================
# 10. คำนวณ MAE
# ==========================================

mae = mean_absolute_error(
    y_test,
    y_pred
)


# ==========================================
# 11. คำนวณ RMSE
# ==========================================

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred
    )
)


# ==========================================
# 12. คำนวณ R²
# ==========================================

r2 = r2_score(
    y_test,
    y_pred
)


# ==========================================
# 13. แสดงผล
# ==========================================

print("\n===================================")
print("Random Forest Regression Results")
print("===================================")

print(f"MAE  : {mae:,.2f}")
print(f"RMSE : {rmse:,.2f}")
print(f"R²   : {r2:.4f}")
```

```python
# ==========================================
# 1. Import Library
# ==========================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


# ==========================================
# 2. อ่านข้อมูล
# ==========================================

df = pd.read_csv("ai_ds_job_salaries_2026.csv")

print("ขนาดข้อมูล:", df.shape)


# ==========================================
# 3. กำหนดตัวแปร X และ Y
# ==========================================

X = df[
    [
        "years_experience",
        "education_level",
        "experience_level",
        "company_size"
    ]
]

y = df["salary_usd"]


# ==========================================
# 4. แบ่งข้อมูล Train / Test
# ใช้ชุดเดียวกันสำหรับทั้ง 2 โมเดล
# ==========================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


# ==========================================
# 5. กำหนดตัวแปรประเภทข้อมูล
# ==========================================

numeric_features = [
    "years_experience"
]

categorical_features = [
    "education_level",
    "experience_level",
    "company_size"
]


# ==========================================
# 6. เตรียมข้อมูล
# ==========================================

numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)

categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)


# ==========================================
# 7. สร้าง Multiple Linear Regression
# ==========================================

model_lr = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("linear_regression", LinearRegression())
    ]
)


# ==========================================
# 8. สร้าง Random Forest Regression
# ==========================================

model_rf = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("random_forest", RandomForestRegressor(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        ))
    ]
)


# ==========================================
# 9. Train Multiple Linear Regression
# ==========================================

model_lr.fit(X_train, y_train)

y_pred_lr = model_lr.predict(X_test)


# ==========================================
# 10. Train Random Forest Regression
# ==========================================

model_rf.fit(X_train, y_train)

y_pred_rf = model_rf.predict(X_test)


# ==========================================
# 11. คำนวณค่าประเมินโมเดล
# ==========================================

# Multiple Linear Regression
mae_lr = mean_absolute_error(y_test, y_pred_lr)

rmse_lr = np.sqrt(
    mean_squared_error(y_test, y_pred_lr)
)

r2_lr = r2_score(y_test, y_pred_lr)


# Random Forest Regression
mae_rf = mean_absolute_error(y_test, y_pred_rf)

rmse_rf = np.sqrt(
    mean_squared_error(y_test, y_pred_rf)
)

r2_rf = r2_score(y_test, y_pred_rf)


# ==========================================
# 12. สร้างตารางเปรียบเทียบ
# ==========================================

comparison = pd.DataFrame({
    "Model": [
        "Multiple Linear Regression",
        "Random Forest Regression"
    ],

    "MAE": [
        mae_lr,
        mae_rf
    ],

    "RMSE": [
        rmse_lr,
        rmse_rf
    ],

    "R²": [
        r2_lr,
        r2_rf
    ]
})


# แสดงผล
display(comparison)
```
